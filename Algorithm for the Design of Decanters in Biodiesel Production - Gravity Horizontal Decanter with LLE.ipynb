```python
from dataclasses import dataclass
from math import pi, sqrt, acos, log, exp
from typing import List

# ============================================================
# LLE SHORTCUT
# ============================================================

@dataclass
class LLEInputs:
    # Order: [Biodiesel (FAME), Alcohol (EtOH), Glycerol]
    z: List[float]         # global mass fractions
    log10K: List[float]    # log10(K_i = heavy / light)
    T: float               # K

@dataclass
class PhaseResult:
    wL: List[float]        # light phase 
    wH: List[float]        # heavy phase 
    phi_L: float

def _to_K(log10K):
    return [10.0**v for v in log10K]

def _norm(w):
    s = sum(max(0.0, wi) for wi in w)
    return [wi/s if s > 0 else 0.0 for wi in w]

def _phi_from(z, wL, wH):
    vals = []
    for zi, xL, xH in zip(z, wL, wH):
        den = xL - xH
        if abs(den) > 1e-12:
            vals.append((zi - xH) / den)
    return max(0.0, min(1.0, sum(vals)/len(vals) if vals else 0.5))

def lle_eq15(inputs: LLEInputs) -> PhaseResult:
    z, K = inputs.z, _to_K(inputs.log10K)
    wL_raw = [2.0 * z[i] / (1.0 + K[i]) for i in range(len(z))]
    wL = _norm(wL_raw)
    wH_raw = [K[i] * wL_raw[i] for i in range(len(z))]
    wH = _norm(wH_raw)
    phi_L = _phi_from(z, wL, wH)
    return PhaseResult(wL=wL, wH=wH, phi_L=phi_L)

# ============================================================
# DATA STRUCTURES
# ============================================================

@dataclass
class FluidProps:
    rho: float      # density of continuous phase (kg/m3)
    mu: float       # dynamic viscosity of continuous phase (Pa·s)

@dataclass
class DispersedProps:
    rho: float      # density of dispersed droplet (kg/m3)
    d_min: float    # target droplet diameter to remove (m)

@dataclass
class DesignTargets:
    Q_m3h: float                          # total volumetric flowrate (m3/h) to the decanter
    safety_u: float = 1.25                # superficial velocity safety factor
    L_over_D: float = 5.0                 # decanter aspect ratio (L/D)
    H_liq_fraction_of_D: float = 0.98     # liquid height 
    api_factor_F: float = 1.0             # API-like area factor
    lamella_multiplier: float = 1.0       # effective area multiplier if lamellas used (>1)

@dataclass
class Geometry:
    D: float           # internal diameter (m)
    L: float           # cylindrical length (m)
    H: float           # liquid height (m)
    A_cross: float     # cross-sectional liquid area (m2) 
    S_surface: float   # plan (horizontal) surface area (m2)

@dataclass
class Checks:
    V_sup: float          # superficial (overflow) velocity Q/S (m/s)
    u_t: float            # terminal velocity for design droplet (m/s)
    V_h: float            # horizontal velocity (m/s)
    t_res: float          # residence time (s)
    weir_loading: float   # Q / L_weir (m3/h per m), if L_weir provided 

# ============================================================
# TERMINAL VELOCITY MODELS
# ============================================================

def terminal_velocity_stokes(mu, drho, d, g=9.81):
    return g * d**2 * drho / (18.0 * mu)

def schiller_naumann_ut(rho, mu, drho, d, g=9.81, tol=1e-9, max_iter=100):
    A = pi * d**2 / 4.0
    Vp = pi * d**3 / 6.0
    u = max(1e-9, terminal_velocity_stokes(mu, drho, d, g))
    for _ in range(max_iter):
        Re = rho * d * u / mu if mu > 0 else 1e6
        Cd = (24.0 / Re) * (1.0 + 0.15 * Re**0.687) if Re > 1e-9 else 1e9
        u_new = sqrt(2.0 * drho * Vp * g / (rho * Cd * A))
        if abs(u_new - u) < tol * max(1.0, u_new):
            return u_new
        u = u_new
    return u

def terminal_velocity_newton_limit(rho, drho, d, g=9.81, Cd_const=0.44):
    return sqrt(4.0 * drho * g * d / (3.0 * rho * Cd_const))

# ============================================================
# DIMENSIONLESS AREA FUNCTION AND GEOMETRY
# ============================================================

def g_f(f):
    """
    Dimensionless wetted-area function (Eq. 32):
      g(f) = 0.25 * [arccos(1 - 2f) - (1 - 2f)*sqrt(f*(1-f)) * 2]
    """
    f = max(1e-9, min(1.0 - 1e-9, f))
    return 0.25 * (acos(1.0 - 2.0*f) - (1.0 - 2.0*f) * 2.0 * sqrt(f * (1.0 - f)))

def chord_at_f(f):
    """Dimensionless chord at height f*D: c = 2*sqrt(f*(1-f)), full chord = D * chord_at_f(f)"""
    f = max(0.0, min(1.0, f))
    return 2.0 * sqrt(f * (1.0 - f))

def circular_segment_area(D, H):
    """Wetted cross-sectional area at height H in cylinder of diameter D."""
    f = max(0.0, min(1.0, H / max(1e-12, D)))
    return D**2 * g_f(f)

def solve_fint_bisection(phi_bottom_vol, f_fill, tol=1e-8, max_iter=200):
    """
    Solve g(fint) = phi_bottom_vol * g(f_fill) using bisection (Eq. 37).
    Returns fint (dimensionless interface height).
    """
    target = phi_bottom_vol * g_f(f_fill)
    lo, hi = 0.0, f_fill
    for _ in range(max_iter):
        fmid = 0.5 * (lo + hi)
        if g_f(fmid) < target:
            lo = fmid
        else:
            hi = fmid
        if (hi - lo) < tol:
            break
    return 0.5 * (lo + hi)

def size_decanter_geometry(S_req, L_over_D, f_fill, phi_bottom_vol):
    """
    Direct (non-iterative) decanter sizing from Eqs. 37–51:
      1. Solve g(fint) = phi_bottom * g(ffill) by bisection → fint
      2. D = sqrt(S_req / (2*alpha*sqrt(fint*(1-fint))))
      3. L, Hliq, Zh, Zl follow directly.
    """
    alpha = max(0.1, float(L_over_D))
    f_fill = max(1e-3, min(1.0 - 1e-3, float(f_fill)))

    fint = solve_fint_bisection(phi_bottom_vol, f_fill)

    # Eq. 47: D from Hazen requirement
    D = sqrt(S_req / max(1e-12, 2.0 * alpha * sqrt(fint * (1.0 - fint))))
    L = alpha * D           # Eq. 48
    H_liq = f_fill * D     # Eq. 49
    Zh = fint * D           # Eq. 50
    Zl = H_liq - Zh        # Eq. 51

    A_cross = D**2 * g_f(f_fill)                    # cross-section at fill level
    S_surface = L * D * chord_at_f(fint)             # Eq. 44: Sgeom = 2*alpha*D^2*sqrt(fint*(1-fint))

    geom = Geometry(D=D, L=L, H=H_liq, A_cross=A_cross, S_surface=S_surface)
    return geom, Zh, Zl, fint

def required_surface_area(Q_m3h, u_t, safety_u=1.25, api_factor_F=1.0, lamella_multiplier=1.0):
    """S_req = Q / V_sup, with V_sup = u_t / safety_u. Applies API factor F and lamella multiplier."""
    Q = Q_m3h / 3600.0
    V_sup = max(1e-12, u_t / max(1.0, safety_u))
    S_api = (Q / V_sup) * max(1.0, api_factor_F)
    return S_api / max(1.0, lamella_multiplier)

def compute_checks(geom, Q_m3h, u_t, L_weir=None):
    Q = Q_m3h / 3600.0
    V_sup = Q / max(1e-12, geom.S_surface)
    V_h = Q / max(1e-12, geom.A_cross)
    t_res = geom.A_cross * geom.L / max(1e-12, Q)
    weir_loading = (Q_m3h / L_weir) if (L_weir and L_weir > 0) else 0.0
    return Checks(V_sup=V_sup, u_t=u_t, V_h=V_h, t_res=t_res, weir_loading=weir_loading)

# ============================================================
# MIXTURE PROPERTIES
# ============================================================

def mass_to_mole_fractions(w, MW_gmol):
    mol_rel = [w[i] / max(1e-16, MW_gmol[i]) for i in range(len(w))]
    s = sum(mol_rel)
    return [mi/s for mi in mol_rel] if s > 0 else [0.0]*len(w)

def mix_molar_density_from_x_and_ci(x, c_i_molm3):
    denom = sum(x[i] / max(1e-16, c_i_molm3[i]) for i in range(len(x)))
    return 1.0 / max(1e-16, denom)

def MW_avg_from_x(x, MW_gmol):
    return sum(x[i] * MW_gmol[i] for i in range(len(x)))

def rho_mass_from_cmix_and_MWavg(c_mix_molm3, MWavg_gmol):
    return c_mix_molm3 * (MWavg_gmol / 1000.0)

def mix_viscosity_biodiesel_phase(w, MW_gmol, mu_pure):
    x = mass_to_mole_fractions(w, MW_gmol)
    ln_mu = sum(x[i] * log(max(1e-16, mu_pure[i])) for i in range(len(w)))
    return exp(ln_mu)

def mix_viscosity_alcoholic_phase(w, MW_gmol, mu_pure, p=0.24):
    x = mass_to_mole_fractions(w, MW_gmol)
    s = sum(x[i] * max(1e-16, mu_pure[i])**p for i in range(len(w)))
    return max(1e-16, s)**(1.0/p)

def mix_viscosity_from_w(w, MW_gmol, mu_pure, rule="auto", alcohol_index=1, thr=0.50):
    if rule == "biodiesel":
        return mix_viscosity_biodiesel_phase(w, MW_gmol, mu_pure)
    if rule == "alcoholic":
        return mix_viscosity_alcoholic_phase(w, MW_gmol, mu_pure)
    x = mass_to_mole_fractions(w, MW_gmol)
    return (mix_viscosity_alcoholic_phase(w, MW_gmol, mu_pure)
            if x[alcohol_index] >= thr
            else mix_viscosity_biodiesel_phase(w, MW_gmol, mu_pure))

# ============================================================
# DECANTER DESIGN FUNCTION
# ============================================================

def design_decanter(fluid, dispersed, targets, method="auto"):
    drho = abs(dispersed.rho - fluid.rho)
    if method == "stokes":
        u_t = terminal_velocity_stokes(fluid.mu, drho, dispersed.d_min)
    elif method == "schiller":
        u_t = schiller_naumann_ut(fluid.rho, fluid.mu, drho, dispersed.d_min)
    elif method == "newton":
        u_t = terminal_velocity_newton_limit(fluid.rho, drho, dispersed.d_min)
    else:
        u_guess = terminal_velocity_stokes(fluid.mu, drho, dispersed.d_min)
        Re = fluid.rho * dispersed.d_min * u_guess / max(1e-12, fluid.mu)
        u_t = u_guess if Re <= 0.5 else schiller_naumann_ut(fluid.rho, fluid.mu, drho, dispersed.d_min)
    S = required_surface_area(targets.Q_m3h, u_t, targets.safety_u,
                               targets.api_factor_F, targets.lamella_multiplier)
    return u_t, S

# ============================================================
# NOZZLE SIZING UTILITIES
# ============================================================

DN_TABLE = [
    ("DN20", 26.9e-3), ("DN25", 33.7e-3), ("DN32", 42.4e-3),
    ("DN40", 48.3e-3), ("DN50", 60.3e-3), ("DN65", 76.1e-3),
    ("DN80", 88.9e-3), ("DN100", 114.3e-3), ("DN125", 139.7e-3),
]

def nozzle_ID(Q_m3h, v_mps):
    return sqrt(4.0 * (Q_m3h / 3600.0) / (pi * max(1e-12, v_mps)))

def to_DN(di):
    for name, od in DN_TABLE:
        if od >= di:
            return name, od
    return DN_TABLE[-1]

# ============================================================
# COST FUNCTION
# ============================================================

def equipment_cost(qE, CB=1.15e4, qB=5.0, b=0.53, fM=1.0, fP=1.0, fT=1.0,
                   CEPCI_current=799.1, CEPCI_base=394.1):
    CE = CB * (qE / max(1e-16, qB))**b * fM * fP * fT
    return CE * (CEPCI_current / CEPCI_base)

# ============================================================
# INPUT DIALOG
# ============================================================

def get_feed_inputs():

    input(
        "WELCOME\n\n"
        "This algorithm is designed for the detailed design of a horizontal\n"
        "liquid-liquid gravity decanter.\n\n"
        "The default feed is defined for a biodiesel production system using:\n\n"
        "  - Biodiesel from palm oil\n"
        "  - Ethanol\n"
        "  - Glycerol\n\n"
        "Temperature: 298.15 K\n\n"
        "If you want to enter your own feed mass flow rates, you will be able\n"
        "to do so in the following input boxes.\n\n"
        "If you prefer to use the default feed values, simply press Enter\n"
        "in each input box.\n\n"
        "Press Enter to continue..."
    )

    def prompt(label, default):
        raw = input(
            f"\n{label}\n"
            f"Default value = {default} kg/h\n"
            "Press Enter to use the default value: "
        ).strip()

        return float(raw) if raw else float(default)

    m_B = prompt("Enter biodiesel (FAEE) mass flow rate", 5396.5)
    m_EtOH = prompt("Enter ethanol (EtOH) mass flow rate", 378.596)
    m_G = prompt("Enter glycerol (Gly) mass flow rate", 565.4)

    return m_B, m_EtOH, m_G

# ============================================================
# MAIN
# ============================================================

if __name__ == "__main__":
    m_B, m_EtOH, m_G = get_feed_inputs()

    T2 = 298.15          # K
    T  = T2-273.15     # °C

    m_tot = m_B + m_EtOH + m_G
    z = [m_B/m_tot, m_EtOH/m_tot, m_G/m_tot]

    # ---- LLE (Palm Oil – EtOH – Glycerol @ 298.15 K)
    PALMOIL_298_LOG10K = [-2.061662, 0.528448, 1.633562]
    lle_res = lle_eq15(LLEInputs(z=z, log10K=PALMOIL_298_LOG10K, T=T2))

    wA, wB = lle_res.wL, lle_res.wH

    # ---- Molecular weights (g/mol)
    MW_gmol = [284.91, 46.068, 92.0938]   # [Biodiesel, EtOH, Glycerol]

    # ---- Pure-component mass densities (kg/m3)
    rho_Bmasica = -1.18*T+895.41
    rho_EtOHmasica = -0.9908*T+810.98
    rho_Gmasica = -0.6914*T+1275.6

    # ---- Pure-component molar densities (mol/m3)
    c_pure = [
        rho_Bmasica/(284.91/1000.0),
        rho_EtOHmasica/(46.068/1000.0),
        rho_Gmasica/(92.0938/1000.0),
    ]

    # ---- Pure-component viscosities (Pa·s)
    mu_B = exp(-0.8677*log(T)+4.6787)*1e-3
    mu_G = exp(-2.889 *log(T)+16.1181)*1e-3
    mu_EtOH = exp(-0.3228*(log(T))**2+1.6477*log(T)-1.8728)*1e-3
    mu_pure = [mu_B, mu_EtOH, mu_G]

    # ---- Phase properties (mole-fraction route)
    xA = mass_to_mole_fractions(wA, MW_gmol)
    xB = mass_to_mole_fractions(wB, MW_gmol)
    rhoA = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xA, c_pure), MW_avg_from_x(xA, MW_gmol))
    rhoB = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xB, c_pure), MW_avg_from_x(xB, MW_gmol))
    muA  = mix_viscosity_from_w(wA, MW_gmol, mu_pure, rule="auto")
    muB  = mix_viscosity_from_w(wB, MW_gmol, mu_pure, rule="auto")

    # ---- Assign Light / Heavy by density
    if rhoA <= rhoB:
        wL, wH = wA, wB
        xL, xH = xA, xB
        rho_L, rho_H = rhoA, rhoB
        mu_L,  mu_H  = muA,  muB
        phiL_mass = lle_res.phi_L
        src_map = "LLE-L → light, LLE-H → heavy"
    else:
        wL, wH = wB, wA
        xL, xH = xB, xA
        rho_L, rho_H = rhoB, rhoA
        mu_L,  mu_H  = muB,  muA
        phiL_mass = 1.0 - lle_res.phi_L
        src_map = "LLE-H → light, LLE-L → heavy (swapped)"

    MW_prom_L = MW_avg_from_x(xL, MW_gmol)
    MW_prom_H = MW_avg_from_x(xH, MW_gmol)
    c_mix_L   = mix_molar_density_from_x_and_ci(xL, c_pure)
    c_mix_H   = mix_molar_density_from_x_and_ci(xH, c_pure)

    # ---- Volumetric flow rates by phase
    phiH_mass = 1.0 - phiL_mass
    Q_L_m3h = (phiL_mass * m_tot)/max(1e-12, rho_L)
    Q_H_m3h = (phiH_mass * m_tot)/max(1e-12, rho_H)

    # ---- Selker–Sleicher criterion: continuous phase selection
    theta = (Q_L_m3h / max(1e-12, Q_H_m3h)) * ((rho_L * mu_H) / (rho_H * mu_L))**0.3
    d_ref = 150e-6
    if theta < 0.5:
        rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L
        cont_label = "heavy (H)"
    elif theta > 2.0:
        rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
        cont_label = "light (L)"
    else:
        u_if_L = terminal_velocity_stokes(mu_L, abs(rho_H - rho_L), d_ref)
        u_if_H = terminal_velocity_stokes(mu_H, abs(rho_L - rho_H), d_ref)
        if u_if_L <= u_if_H:
            rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
            cont_label = "light (L, conservative)"
        else:
            rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L
            cont_label = "heavy (H, conservative)"

    # ---- Total volumetric flow to decanter
    Q_total_m3h = m_B/rho_Bmasica + m_EtOH/rho_EtOHmasica + m_G/rho_Gmasica
    d_target = 150e-6

    targets = DesignTargets(
        Q_m3h=Q_total_m3h, safety_u=1.25, L_over_D=5.0,
        H_liq_fraction_of_D=0.98, api_factor_F=1.0, lamella_multiplier=1.0
    )
    fluid     = FluidProps(rho=rho_cont, mu=mu_cont)
    dispersed = DispersedProps(rho=rho_disp, d_min=d_target)

    u_t, S = design_decanter(fluid, dispersed, targets, method="auto")

    # ---- Phase volumetric fractions for interface calculation
    Q_sum = max(1e-12, Q_L_m3h + Q_H_m3h)
    phiL_vol = Q_L_m3h / Q_sum
    phiH_vol = 1.0 - phiL_vol
    bottom_is_H = (rho_H >= rho_L)
    phi_bottom_vol = phiH_vol if bottom_is_H else phiL_vol

    # ---- Direct geometry sizing (no iteration on fint initial guess)
    f_fill = targets.H_liq_fraction_of_D
    geom, Zh, Zl, fint = size_decanter_geometry(
        S_req=S, L_over_D=targets.L_over_D,
        f_fill=f_fill, phi_bottom_vol=phi_bottom_vol
    )

    # ---- Gepm
    chk = compute_checks(geom, targets.Q_m3h, u_t)
    geom = Geometry(D=geom.D, L=geom.L, H=geom.H,
                        A_cross=geom.A_cross, S_surface=geom.S_surface)
    chk = compute_checks(geom, targets.Q_m3h, u_t)

    # ---- Phase areas and labels
    A_total    = geom.A_cross
    A_fondo    = phi_bottom_vol * A_total
    A_superior = A_total - A_fondo

    def _rich_label(w):
        names = ["biodiesel", "ethanol", "glycerol"]
        return names[max(range(len(w)), key=lambda k: w[k])]

    fase_fondo = f"{'H' if bottom_is_H else 'L'} (rich in {_rich_label(wH if bottom_is_H else wL)})"

    # ---- Nozzle sizing
    v_feed, v_out_L, v_out_H = 0.30, 0.50, 0.50
    D_inlet_ID = nozzle_ID(Q_total_m3h, v_feed)
    D_Lout_ID  = nozzle_ID(Q_L_m3h, v_out_L)
    D_Hout_ID  = nozzle_ID(Q_H_m3h, v_out_H)
    DN_inlet, DN_inlet_OD = to_DN(D_inlet_ID)
    DN_Lout,  DN_Lout_OD  = to_DN(D_Lout_ID)
    DN_Hout,  DN_Hout_OD  = to_DN(D_Hout_ID)

    # ---- Internal baffles
    from math import ceil
    dispersion_band           = 0.10 * geom.D
    inlet_to_first_baffle     = 0.275 * geom.D
    partial_baffle_height     = 0.46
    main_baffle_gap           = 0.05
    main_baffle_height        = max(0.0, Zh - main_baffle_gap)
    main_baffle_hole_d        = 0.025
    main_baffle_perforated_band = 0.152
    # Perforated baffle
    hole_d = 0.015
    A_nozzles_out = (pi/4.0) * (D_Lout_ID**2 + D_Hout_ID**2)
    N_holes = max(1, ceil(A_nozzles_out / (pi*(hole_d/2.0)**2)))

    # ---- Volume and cost
    Vol = geom.A_cross * geom.L
    qE  = (pi * geom.D**2 / 4.0) * geom.L
    CE_final = equipment_cost(qE)

    # ---- Reynolds of design droplet
    Re = fluid.rho * dispersed.d_min * u_t / max(1e-16, fluid.mu)
    Vsup_obj = u_t / targets.safety_u

    # ============================================================
    # REPORTS
    # ============================================================
    print("=" * 70)
    print("  PHASE PROPERTIES")
    print("=" * 70)
    print(f"[LIGHT] x={['%.5f'%xi for xi in xL]}, MW={MW_prom_L:.2f} g/mol, "
          f"rho={rho_L:.2f} kg/m3, mu={mu_L:.4e} Pa·s")
    print(f"[HEAVY] x={['%.5f'%xi for xi in xH]}, MW={MW_prom_H:.2f} g/mol, "
          f"rho={rho_H:.2f} kg/m3, mu={mu_H:.4e} Pa·s")
    print(f"wL (light, mass) = {['%.10f'%wi for wi in wL]}")
    print(f"wH (heavy, mass) = {['%.10f'%wi for wi in wH]}")
    print(f"c_mix_L/H (mol/m3) = {c_mix_L:.1f}, {c_mix_H:.1f}")
    print(f"Pure densities: rho_B={rho_Bmasica:.4f}, rho_EtOH={rho_EtOHmasica:.4f}, "
          f"rho_G={rho_Gmasica:.4f} kg/m3")
    print(f"Pure viscosities: mu_B={mu_B:.4e}, mu_EtOH={mu_EtOH:.4e}, mu_G={mu_G:.4e} Pa·s")
    print(f"Phase mapping: {src_map}")

    print(f"\nθ (Selker–Sleicher) = {theta:.8f}")
    print(f"Q_L = {Q_L_m3h:.2f} m³/h  |  Q_H = {Q_H_m3h:.2f} m³/h")
    print(f"Continuous phase → {cont_label}")
    print(f"Continuous: rho={rho_cont:.2f} kg/m3, mu={mu_cont:.4e} Pa·s  |  "
          f"Dispersed: rho={rho_disp:.2f} kg/m3")

    print("\n" + "=" * 70)
    print("  INTERFACE POSITION (Hazen at interface + volumetric phase split)")
    print("=" * 70)
    print(f"ρ_L = {rho_L:.2f} kg/m³,  ρ_H = {rho_H:.2f} kg/m³  →  Bottom phase: {fase_fondo}")
    print(f"u_t (terminal) = {u_t:.6e} m/s  |  Re = {Re:.6f}")
    print(f"φ_L (mass) = {phiL_mass:.4f}  |  φ_L (vol) = {phiL_vol:.4f}")
    print(f"S_req (Hazen)  = {S:.3f} m²")
    print(f"f_fill = {f_fill:.4f}  |  f_int (bisection) = {fint:.4f}")
    print(f"D = {geom.D:.3f} m  |  L = {geom.L:.3f} m  |  H_liq = {geom.H:.3f} m")
    print(f"Zh = {Zh:.3f} m  (Zh/D = {Zh/geom.D:.4f})  |  Zl = {Zl:.3f} m  (Zh+Zl = H_liq)")
    print(f"S_geom = L*c(D,Zh) = {geom.S_surface:.3f} m²  (should ≈ S_req)")
    print(f"A_cross total = {A_total:.4f} m²  →  A_L = {A_superior:.4f} m²  |  A_H = {A_fondo:.4f} m²")
    print(f"V_sup = {chk.V_sup:.6e} m/s")
    print(f"V_h   = {chk.V_h:.6e} m/s")
    print(f"t_res = {chk.t_res/60:.2f} min")
    print(f"Q_total = {Q_total_m3h:.4f} m³/h")
    print(f"Vol (useful) = {Vol:.3f} m³  |  H/D = {geom.H/geom.D:.3f}")

    print("\n" + "=" * 70)
    print("  INTERNALS AND NOZZLES")
    print("=" * 70)
    print(f"Dispersion band = {dispersion_band:.3f} m")
    print(f"Perforated baffle location: {inlet_to_first_baffle:.3f} m above decanter bottom")
    print(f"Perforated baffle height: {geom.D - inlet_to_first_baffle:.3f} m")
    print(f"Perforated baffle: Ø hole = {hole_d*1000:.0f} mm  |  N° holes ≈ {N_holes}  "
          f"(holes area ≈ outlet nozzle areas)")
    print(f"Main solid baffle height ≈ Zh − {main_baffle_gap*1000:.0f} mm = {main_baffle_height:.3f} m")
    print(f"Main solid baffle: Ø holes = {main_baffle_hole_d*1000:.0f} mm, "
          f"only in upper {main_baffle_perforated_band*1000:.0f} mm perforated band")
    print(f"Partial baffle height ≈ {partial_baffle_height:.3f} m")
    print(f"Oulet perforated baffle height ≈ {Zh-0.15:.3f} m")
    print(f"\n-- Nozzles (calculated ID → suggested DN) --")
    print(f"Feed:  D_ID = {D_inlet_ID*1000:.1f} mm  →  {DN_inlet} (~{DN_inlet_OD*1000:.1f} mm)  @ v ≈ {v_feed:.2f} m/s")
    print(f"Light: D_ID = {D_Lout_ID*1000:.1f} mm  →  {DN_Lout}  (~{DN_Lout_OD*1000:.1f} mm)  @ v ≈ {v_out_L:.2f} m/s")
    print(f"Heavy: D_ID = {D_Hout_ID*1000:.1f} mm  →  {DN_Hout}  (~{DN_Hout_OD*1000:.1f} mm)  @ v ≈ {v_out_H:.2f} m/s")
    print(f"Vent:  DN20 (~26.9 mm)  (typical selection)")
    print(f"Drain: DN50 (~50.8 mm)  (flush with floor)")
    print("NOTE: Approximate DNs with OD as ID proxy; confirm with Schedule")

    # ---- Design notes
    notes = []
    if chk.V_sup > Vsup_obj * 1.02:
        notes.append("V_sup > target (review S, safety_u or u_t).")
    if chk.V_h > 0.02:
        notes.append("V_h high (consider increasing D or parallel units).")
    print("\nNotes:", "Preliminary design within specified criteria." if not notes else "")
    for n in notes:
        print(f"  - {n}")

    print("\n" + "=" * 70)
    print("  MASS SPLIT BY PHASE (kg/h)")
    print("=" * 70)
    comp_names = ["Biodiesel", "Ethanol", "Glycerol"]
    m_feed  = [m_B, m_EtOH, m_G]
    m_L     = phiL_mass * m_tot
    m_H     = phiH_mass * m_tot
    mL_i    = [m_L * wi for wi in wL]
    mH_i    = [m_H * wi for wi in wH]
    print(f"Feed total = {m_tot:.3f} kg/h  |  φ_L (mass) = {phiL_mass:.4f}")
    print(f"Bottom phase = {'Heavy' if bottom_is_H else 'Light'} "
          f"(ρ_H = {rho_H:.1f} vs ρ_L = {rho_L:.1f} kg/m³)")
    print("\nLight phase:")
    for nm, val in zip(comp_names, mL_i):
        print(f"  {nm:<10s} = {val:,.3f}")
    print(f"  Total_L    = {m_L:,.3f}")
    print("\nHeavy phase:")
    for nm, val in zip(comp_names, mH_i):
        print(f"  {nm:<10s} = {val:,.3f}")
    print(f"  Total_H    = {m_H:,.3f}")

    try:
        import pandas as pd
        df_mass = pd.DataFrame({
            "Component":    comp_names + ["TOTAL"],
            "Light (kg/h)": mL_i + [m_L],
            "Heavy (kg/h)": mH_i + [m_H],
            "Feed (kg/h)":  m_feed + [m_tot],
        })
        try:
            from caas_jupyter_tools import display_dataframe_to_user
            display_dataframe_to_user("Decanter Mass Split (kg/h)", df_mass)
        except Exception:
            print("\n", df_mass.to_string(index=False))
    except Exception:
        pass

    print("\n" + "=" * 70)
    print("  EQUIPMENT COST (capacity rule + CEPCI)")
    print("=" * 70)
    print(f"Volume qE        = {qE:.3f} m³")
    print(f"Base CB          = {1.15e4:,.2f}  @ qB=5 m³,  b=0.53,  fM=fP=fT=1")
    CE_unindexed = 1.15e4 * (qE / 5.0)**0.53
    print(f"CE (unindexed)   = {CE_unindexed:,.2f}")
    print(f"CEPCI factor     = {799.1/394.1:.4f}  (799.1 / 394.1)")
    print(f"CE (final)       = {CE_final:,.2f} USD")

# ============================================================
# SENSITIVITY ANALYSIS – DROPLET DIAMETER
# ============================================================

# %%
# %matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from math import pi, log, exp, sqrt, acos

m_B_base    = float(globals().get("m_B",    5396.5))
m_EtOH_base = float(globals().get("m_EtOH", 378.596))
m_G_base    = float(globals().get("m_G",    565.4))
print(f"Base case: m_B={m_B_base} | m_EtOH={m_EtOH_base} | m_G={m_G_base} kg/h")

def _eval_with_d(m_B_val, m_EtOH_val, m_G_val, d_target_m=150e-6):
    T2 = 298.15; T = T2 - 273.15
    m_B, m_EtOH, m_G = float(m_B_val), float(m_EtOH_val), float(m_G_val)
    m_tot = m_B + m_EtOH + m_G
    z = [m_B/m_tot, m_EtOH/m_tot, m_G/m_tot]

    lle_res = lle_eq15(LLEInputs(z=z, log10K=[-2.061662, 0.528448, 1.633562], T=T2))
    wA, wB = lle_res.wL, lle_res.wH
    MW_gmol = [284.91, 46.068, 92.0938]

    rho_Bm  = -1.18*T + 895.41
    rho_Em  = -0.9908*T + 810.98
    rho_Gm  = -0.6914*T + 1275.6
    c_pure = [rho_Bm/(284.91/1000), rho_Em/(46.068/1000), rho_Gm/(92.0938/1000)]
    mu_B   = exp(-0.8677*log(T) + 4.6787) * 1e-3
    mu_G   = exp(-2.889*log(T) + 16.1181) * 1e-3
    mu_EtOH = exp(-0.3228*(log(T))**2 + 1.6477*log(T) - 1.8728) * 1e-3
    mu_pure = [mu_B, mu_EtOH, mu_G]

    xA = mass_to_mole_fractions(wA, MW_gmol)
    xB = mass_to_mole_fractions(wB, MW_gmol)
    rhoA = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xA, c_pure), MW_avg_from_x(xA, MW_gmol))
    rhoB = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xB, c_pure), MW_avg_from_x(xB, MW_gmol))
    muA = mix_viscosity_from_w(wA, MW_gmol, mu_pure, rule="auto")
    muB = mix_viscosity_from_w(wB, MW_gmol, mu_pure, rule="auto")

    if rhoA <= rhoB:
        rho_L, rho_H, mu_L, mu_H, phiL_mass = rhoA, rhoB, muA, muB, lle_res.phi_L
    else:
        rho_L, rho_H, mu_L, mu_H, phiL_mass = rhoB, rhoA, muB, muA, 1.0 - lle_res.phi_L

    Q_L = (phiL_mass * m_tot) / max(1e-12, rho_L)
    Q_H = ((1.0 - phiL_mass) * m_tot) / max(1e-12, rho_H)
    theta = (Q_L / max(1e-12, Q_H)) * ((rho_L * mu_H) / (rho_H * mu_L))**0.3

    if theta < 0.5:
        rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L
    elif theta > 2.0:
        rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
    else:
        uL = terminal_velocity_stokes(mu_L, abs(rho_H - rho_L), d_target_m)
        uH = terminal_velocity_stokes(mu_H, abs(rho_L - rho_H), d_target_m)
        if uL <= uH:
            rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
        else:
            rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L

    Q_total = m_B/rho_Bm + m_EtOH/rho_Em + m_G/rho_Gm
    drho = abs(rho_disp - rho_cont)
    u_guess = terminal_velocity_stokes(mu_cont, drho, d_target_m)
    Re_g = rho_cont * d_target_m * u_guess / max(1e-12, mu_cont)
    u_t = u_guess if Re_g <= 0.5 else schiller_naumann_ut(rho_cont, mu_cont, drho, d_target_m)

    S = required_surface_area(Q_m3h=Q_total, u_t=u_t, safety_u=1.25)
    phiL_vol = Q_L / max(1e-12, Q_L + Q_H)
    phi_bottom_vol = (1.0 - phiL_vol) if (rho_H >= rho_L) else phiL_vol
    geom, Zh, Zl, fint = size_decanter_geometry(S_req=S, L_over_D=5.0, f_fill=0.98, phi_bottom_vol=phi_bottom_vol)

    chk = compute_checks(geom, Q_total, u_t)
    geom = Geometry(D=geom.D, L=geom.L, H=geom.H,
                        A_cross=geom.A_cross, S_surface=geom.S_surface)
    chk = compute_checks(geom, Q_total, u_t)

    Vol = pi * (geom.D/2)**2 * geom.L
    qE = (pi * geom.D**2 / 4.0) * geom.L
    CE_final = equipment_cost(qE)
    return {"S_m2": S, "D_m": geom.D, "L_m": geom.L, "Vol_m3": Vol,
            "t_res_min": chk.t_res/60.0, "CE_USD": CE_final, "u_t": u_t}

d_um_list = [50, 75, 100, 125, 150, 200, 250, 300]
filas = []
for d_um in d_um_list:
    try:
        r = _eval_with_d(m_B_base, m_EtOH_base, m_G_base, d_target_m=d_um*1e-6)
        filas.append({"d_min (µm)": d_um, "u_t (m/s)": round(r["u_t"], 6),
                      "S_Hazen (m²)": round(r["S_m2"], 3), "D (m)": round(r["D_m"], 3),
                      "L (m)": round(r["L_m"], 3), "Vol (m³)": round(r["Vol_m3"], 3),
                      "t_res (min)": round(r["t_res_min"], 2), "CE (USD)": round(r["CE_USD"], 0)})
    except Exception as e:
        print(f"  Error d={d_um} µm: {e}")

df_d = pd.DataFrame(filas)
display(df_d.style.set_caption("Droplet diameter sensitivity – base case")
        .format({"d_min (µm)": "{:.0f}", "u_t (m/s)": "{:.6f}", "S_Hazen (m²)": "{:.3f}",
                 "D (m)": "{:.3f}", "L (m)": "{:.3f}", "Vol (m³)": "{:.3f}",
                 "t_res (min)": "{:.2f}", "CE (USD)": "{:,.0f}"})
        .background_gradient(subset=["S_Hazen (m²)", "D (m)", "CE (USD)"], cmap="YlOrRd"))

x = df_d["d_min (µm)"].to_numpy()
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
pares = [
    (axes[0, 0], "S_Hazen (m²)", "S_Hazen (m²)"),
    (axes[0, 1], "D (m)",        "D (m)"),
    (axes[0, 2], "L (m)",        "L (m)"),
    (axes[1, 0], "Vol (m³)",     "Vol (m³)"),
    (axes[1, 1], "t_res (min)",  "t_res (min)"),
    (axes[1, 2], "CE (USD)",     "CE (USD)"),
]

panel_labels = ["a.", "b.", "c.", "d.", "e.", "f."]

for label, (ax, col, ylab) in zip(panel_labels, pares):
    ax.plot(x, df_d[col].to_numpy(), marker="o", linewidth=2, color="#1f77b4")
    ax.set_xlabel("d_min (µm)", fontsize=9)
    ax.set_ylabel(ylab, fontsize=9)
    ax.set_title(label, fontsize=12, fontweight="bold")
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

base_row = df_d[df_d["d_min (µm)"] == 150]
if not base_row.empty:
    fig3, ax3 = plt.subplots(figsize=(9, 5))
    cols_rel = ["S_Hazen (m²)", "D (m)", "CE (USD)"]
    colors   = ["#1f77b4", "#ff7f0e", "#d62728"]
    for col, color in zip(cols_rel, colors):
        base_val = base_row[col].values[0]
        if base_val != 0:
            pct = 100 * (df_d[col] / base_val - 1)
            ax3.plot(x, pct, marker="o", linewidth=2, color=color, label=col)
    ax3.axhline(0, color="gray", linestyle="--", linewidth=1)
    ax3.set_xlabel("d_min (µm)", fontsize=10)
    ax3.set_ylabel("Variation respect results to d_min=150 µm (%)", fontsize=10)
    ax3.set_title("")
    ax3.legend(fontsize=9)
    ax3.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# SENSITIVITY ANALYSIS – ETHANOL:OIL MOLAR RATIO
# ============================================================

# %%
# %matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
from math import pi, log, exp
from IPython.display import display

CASOS_D100 = [
    ("3.8:1",  249.707604881051), ("4:1",  325.70473466493314), ("5:1",  651.4094693298663),
    ("6:1",  977.1142039947995),  ("7:1", 1302.8189386597326),  ("8:1", 1628.5236733246657),
    ("9:1", 1954.228407989599),   ("10:1", 2279.933142654532),  ("11:1", 2605.637877319465),
    ("12:1", 2931.3426119843984), ("13:1", 3257.0473466493313), ("14:1", 3582.7520813142646),
    ("15:1", 3908.456815979198),  ("16:1", 4234.161550644131),  ("17:1", 4559.866285309064),
    ("18:1", 4885.571019973998),
]
CASOS_D101 = [
    ("4:1",   34.30190284836307), ("5:1",  68.60380569672614),  ("6:1",  102.90570854508921),
    ("7:1",  137.20761139345228), ("8:1",  171.50951424181537), ("9:1",  205.81141709017842),
    ("10:1", 240.1133199385415),  ("11:1", 274.41522278690456), ("12:1", 308.7171256352677),
    ("13:1", 343.01902848363073), ("14:1", 377.3209313319938),  ("15:1", 411.62283418035685),
    ("16:1", 445.92473702871996), ("17:1", 480.226639877083),   ("18:1", 514.5285427254461),
]

def evaluar_decantador(m_B_val, m_EtOH_val, m_G_val):
    T2 = 298.15; T = T2 - 273.15
    m_B, m_EtOH, m_G = float(m_B_val), float(m_EtOH_val), float(m_G_val)
    m_tot_feed = m_B + m_EtOH + m_G
    z = [m_B/m_tot_feed, m_EtOH/m_tot_feed, m_G/m_tot_feed]

    lle_res = lle_eq15(LLEInputs(z=z, log10K=[-2.061662, 0.528448, 1.633562], T=T2))
    wA, wB = lle_res.wL, lle_res.wH
    MW_gmol = [284.91, 46.068, 92.0938]

    rho_Bm  = -1.18*T + 895.41
    rho_Em  = -0.9908*T + 810.98
    rho_Gm  = -0.6914*T + 1275.6
    c_pure = [rho_Bm/(284.91/1000), rho_Em/(46.068/1000), rho_Gm/(92.0938/1000)]
    mu_B   = exp(-0.8677*log(T) + 4.6787) * 1e-3
    mu_G   = exp(-2.889*log(T) + 16.1181) * 1e-3
    mu_EtOH = exp(-0.3228*(log(T))**2 + 1.6477*log(T) - 1.8728) * 1e-3
    mu_pure = [mu_B, mu_EtOH, mu_G]

    xA = mass_to_mole_fractions(wA, MW_gmol)
    xB = mass_to_mole_fractions(wB, MW_gmol)
    rhoA = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xA, c_pure), MW_avg_from_x(xA, MW_gmol))
    rhoB = rho_mass_from_cmix_and_MWavg(mix_molar_density_from_x_and_ci(xB, c_pure), MW_avg_from_x(xB, MW_gmol))
    muA = mix_viscosity_from_w(wA, MW_gmol, mu_pure, rule="auto")
    muB = mix_viscosity_from_w(wB, MW_gmol, mu_pure, rule="auto")

    if rhoA <= rhoB:
        rho_L, rho_H, mu_L, mu_H, phiL_mass = rhoA, rhoB, muA, muB, lle_res.phi_L
    else:
        rho_L, rho_H, mu_L, mu_H, phiL_mass = rhoB, rhoA, muB, muA, 1.0 - lle_res.phi_L

    Q_L_m3h = (phiL_mass * m_tot_feed) / max(1e-12, rho_L)
    Q_H_m3h = ((1.0 - phiL_mass) * m_tot_feed) / max(1e-12, rho_H)
    theta = (Q_L_m3h / max(1e-12, Q_H_m3h)) * ((rho_L * mu_H) / (rho_H * mu_L))**0.3
    d_ref = 150e-6

    if theta < 0.5:
        rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L
    elif theta > 2.0:
        rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
    else:
        uL = terminal_velocity_stokes(mu_L, abs(rho_H - rho_L), d_ref)
        uH = terminal_velocity_stokes(mu_H, abs(rho_L - rho_H), d_ref)
        if uL <= uH:
            rho_cont, mu_cont, rho_disp = rho_L, mu_L, rho_H
        else:
            rho_cont, mu_cont, rho_disp = rho_H, mu_H, rho_L

    Q_total_m3h = m_B/rho_Bm + m_EtOH/rho_Em + m_G/rho_Gm
    drho = abs(rho_disp - rho_cont)
    u_guess = terminal_velocity_stokes(mu_cont, drho, d_ref)
    Re_g = rho_cont * d_ref * u_guess / max(1e-12, mu_cont)
    u_t = u_guess if Re_g <= 0.5 else schiller_naumann_ut(rho_cont, mu_cont, drho, d_ref)

    S = required_surface_area(Q_m3h=Q_total_m3h, u_t=u_t, safety_u=1.25)
    phiL_vol = Q_L_m3h / max(1e-12, Q_L_m3h + Q_H_m3h)
    phi_bottom_vol = (1.0 - phiL_vol) if (rho_H >= rho_L) else phiL_vol
    geom, Zh, Zl, fint = size_decanter_geometry(S_req=S, L_over_D=5.0, f_fill=0.98, phi_bottom_vol=phi_bottom_vol)

    chk = compute_checks(geom, Q_total_m3h, u_t)
    geom = Geometry(D=geom.D, L=geom.L, H=geom.H,
                        A_cross=geom.A_cross, S_surface=geom.S_surface)
    chk = compute_checks(geom, Q_total_m3h, u_t)

    Vol = pi * (geom.D/2)**2 * geom.L
    qE = (pi * geom.D**2 / 4.0) * geom.L
    CE_final = equipment_cost(qE)

    def _rich(w):
        return ["biodiesel", "ethanol", "glycerol"][max(range(len(w)), key=lambda i: w[i])]

    bottom_is_H = (rhoA <= rhoB)
    fase_fondo = f"{'H' if bottom_is_H else 'H'} (rich in {_rich(wB if bottom_is_H else wA)})"

    return {"S_m2": S, "D_m": geom.D, "L_m": geom.L, "H_m": geom.H, "Zh_m": Zh,
            "Vol_m3": Vol, "t_res_min": chk.t_res/60.0, "rho_L": rho_L, "rho_H": rho_H,
            "Q_total": Q_total_m3h, "CE_USD": CE_final, "theta": theta, "fase_fondo": fase_fondo}

def ejecutar_analisis(casos, m_B_fixed, m_G_fixed, titulo):
    print(f"\n{'='*70}\n  {titulo}  |  m_B = {m_B_fixed} kg/h  |  m_G = {m_G_fixed} kg/h\n{'='*70}\n")
    filas = []
    for label, m_EtOH_val in casos:
        try:
            r = evaluar_decantador(m_B_fixed, m_EtOH_val, m_G_fixed)
            filas.append({
                "Molar ratio":      label,
                "m_EtOH (kg/h)":   round(m_EtOH_val, 2),
                "S_Hazen (m²)":    round(r["S_m2"], 3),
                "D (m)":           round(r["D_m"], 3),
                "L (m)":           round(r["L_m"], 3),
                "H_liq (m)":       round(r["H_m"], 3),
                "Zh (m)":          round(r["Zh_m"], 3),
                "Vol (m³)":        round(r["Vol_m3"], 3),
                "t_res (min)":     round(r["t_res_min"], 2),
                "ρ_L (kg/m³)":     round(r["rho_L"], 2),
                "ρ_H (kg/m³)":     round(r["rho_H"], 2),
                "Q_total (m³/h)":  round(r["Q_total"], 3),
                "CE (USD)":        round(r["CE_USD"], 0),
                "θ (Selker–Sleicher)": float(r["theta"]),
                "Bottom phase":    r["fase_fondo"],
            })
        except Exception as e:
            print(f"  !! Error in case {label}: {e}")

    if not filas:
        print("  No results generated.")
        return pd.DataFrame()

    df = pd.DataFrame(filas)
    display(df.style.set_caption(titulo)
            .format({"m_EtOH (kg/h)": "{:,.2f}", "S_Hazen (m²)": "{:,.3f}",
                     "D (m)": "{:.3f}", "L (m)": "{:.3f}", "H_liq (m)": "{:.3f}",
                     "Zh (m)": "{:.3f}", "Vol (m³)": "{:,.3f}", "t_res (min)": "{:,.2f}",
                     "ρ_L (kg/m³)": "{:.2f}", "ρ_H (kg/m³)": "{:.2f}",
                     "Q_total (m³/h)": "{:.3f}", "CE (USD)": "{:,.0f}",
                     "θ (Selker–Sleicher)": "{:.4f}"})
            .background_gradient(subset=["S_Hazen (m²)", "D (m)", "CE (USD)"], cmap="YlOrRd"))

    x_labels = df["Molar ratio"].tolist()
    x_pos = list(range(len(x_labels)))

    fig, axes = plt.subplots(2, 3, figsize=(16, 10))
    pares = [
        (axes[0, 0], "S_Hazen (m²)", "S (m²)",      "#1f77b4"),
        (axes[0, 1], "D (m)",         "D (m)",        "#ff7f0e"),
        (axes[0, 2], "L (m)",         "L (m)",        "#2ca02c"),
        (axes[1, 0], "Vol (m³)",      "Vol (m³)",     "#9467bd"),
        (axes[1, 1], "t_res (min)",   "t_res (min)",  "#8c564b"),
        (axes[1, 2], "CE (USD)",      "CE (USD)",     "#d62728"),
    ]

    panel_labels = ["a.", "b.", "c.", "d.", "e.", "f."," "]
    
    for panel_labels, (ax, col, ylabel, color) in zip(panel_labels, pares):
        ax.plot(x_pos, df[col].tolist(), marker='o', color=color, linewidth=2, markersize=6)
        ax.set_xticks(x_pos)
        ax.set_xticklabels(x_labels, rotation=45, ha='right', fontsize=8)
        ax.set_title(" ")
        ax.set_xlabel("Molar ratio ethanol:oil", fontsize=9)
        ax.set_ylabel(ylabel, fontsize=9)
        ax.grid(True, alpha=0.3)

        if col == "CE (USD)":
           ax.ticklabel_format(axis='y', style='sci', scilimits=(6, 6), useMathText=False)
    plt.tight_layout()
    plt.show()

    fig2, ax2 = plt.subplots(figsize=(9, 5))
    ax2.plot(x_pos, df["ρ_L (kg/m³)"].tolist(), marker='o', color="#1f77b4",
             linewidth=2, markersize=6, label="ρ light phase")
    ax2.plot(x_pos, df["ρ_H (kg/m³)"].tolist(), marker='s', color="#d62728",
             linewidth=2, markersize=6, label="ρ heavy phase")
    ax2.set_xticks(x_pos)
    ax2.set_xticklabels(x_labels, rotation=45, ha='right', fontsize=9)
    ax2.set_title(" ")
    ax2.set_xlabel("Molar ratio ethanol:oil", fontsize=10)
    ax2.set_ylabel("Density (kg/m³)", fontsize=10)
    ax2.legend(fontsize=9)
    ax2.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    return df

print("=" * 70)
print("  SENSITIVITY ANALYSIS – DECANTERS D100 and D101")
print("  Variation of ethanol:oil molar ratio")
print("=" * 70)

df_D100 = ejecutar_analisis(CASOS_D100, m_B_fixed=5396.5, m_G_fixed=565.4, titulo="Decanter D100")
df_D101 = ejecutar_analisis(CASOS_D101, m_B_fixed=6005.5,  m_G_fixed=64,  titulo="Decanter D101")

print("\n" + "=" * 70)
print("  Analysis completed.")
print("=" * 70)
```
